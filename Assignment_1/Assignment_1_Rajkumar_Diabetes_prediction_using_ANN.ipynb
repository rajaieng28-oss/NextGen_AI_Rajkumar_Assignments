```python
# Separate Input Features and Target Output
x=data.drop('Outcome',axis=1)
y=data['Outcome']
```

```python
# Train-Test Data split from Dataset
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# Apply Feature Scaling using StandardScaler
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train_scaled=sc.fit_transform(x_train)
x_test_scaled=sc.transform(x_test)
```

```python
x_train_scaled
```

```python
x_test_scaled
```

```python
# Import required tensorflow/ Keras libraries
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense
```

```python
# Create a sequential Model
model=Sequential()
```

```python
#Add layers such as Input, First Hidden, Second Hidden & output with activation functions
model=Sequential([
    Input(shape=(x_train_scaled.shape[1],),name='input_layer'), #input_layer has 8 neurons
    Dense(32,activation='relu',name='hidden_layer_1'), #hidden_layer_1 has 16 neurons
    Dense(16,activation='relu',name='hidden_layer_2'), #hidden_layer_2 has 32 neurons
    Dense(1,activation='sigmoid',name='output_layer') #output_layer has just 1 neuron since it is a binary classifier
])
```

```python
# Compile the model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
# Train the model
trained_model = model.fit(
    x_train_scaled,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
final_accuracy = trained_model.history['accuracy'][-1]
final_accuracy
```

```python
# Generate Predictions on test Dataset
y_pred=model.predict(x_test_scaled)
```

```python
print(y_pred)
```

```python
# Convert Prediction Probabilities into Binary class labels
y_pred_binary = (y_pred > 0.5).astype(int)
```

```python
print(y_pred_binary)
```

```python
# Evaluate the model
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
# accuracy_score
accuracy = accuracy_score(y_test, y_pred_binary)
print("Accuracy:", accuracy)
```

```python
# confusion_matrix
cm = confusion_matrix(y_test, y_pred_binary)
print("Confusion Matrix:", cm)
```

```python
# Visualizing the confusion matrix
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```
